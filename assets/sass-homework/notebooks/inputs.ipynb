## Practice — local signup preview

The separate extension connects styled input controls to visible output and native form validation.

````python
%%html
<!-- UI_RUNNER: Practice — local signup preview -->
<form id="signup">
<label for="category">Assistance Category</label>
<input id="category" name="category" type="text" class="ocs__input large gradient" placeholder="Example: transportation" required>
<label for="first-name">Volunteer First Name</label>
<input id="first-name" name="firstName" type="text" class="ocs__input small" placeholder="Example: Alex" autocomplete="given-name" required>
<label for="email">Volunteer Email</label>
<input id="email" name="email" type="email" class="ocs__input medium" placeholder="alex@example.com" autocomplete="email" required>
<button class="ocs__btn accent fill" type="submit">Preview signup</button>
</form>
<p id="signup-result" role="status">Enter fictional details to preview a signup.</p>
<script>
document.getElementById('signup').addEventListener('submit', event => {
  event.preventDefault();
  const name = document.getElementById('first-name').value.trim();
  const category = document.getElementById('category').value.trim();
  document.getElementById('signup-result').textContent = name && category
    ? `${name}: your ${category} signup is previewed locally. Nothing was sent.`
    : 'Please enter a name and assistance category, not only spaces.';
});
</script>
````

## Practice — values, conversion, and validation

The resource field updates as you type. The skill button reads a value on click. Age demonstrates string concatenation versus numeric addition; blank and invalid values are handled separately. Email checks illustrate a simple format check, not proof that an address exists.

````python
%%html
<!-- UI_RUNNER: Practice — values, conversion, and validation -->
<label for="resource">Resource need (large gradient variant)</label>
<input id="resource" class="ocs__input large gradient" placeholder="Describe a resource need">
<p id="resource-result" role="status">No resource entered.</p>
<label for="skill">Volunteer skill (compact variant)</label>
<input id="skill" class="ocs__input small" placeholder="Example: organizing">
<button id="skill-submit" type="button" class="ocs__btn accent fill">Show skill</button>
<p id="skill-result" role="status">No skill entered.</p>
<label for="age">Sample age</label>
<input id="age" class="ocs__input medium" type="number" min="0" value="18">
<p id="age-result" role="status"></p>
<label for="practice-email">Sample email</label>
<input id="practice-email" class="ocs__input medium" type="email" placeholder="alex@example.com">
<p id="email-result" role="status"></p>
<script>
const resource = document.getElementById('resource');
resource.addEventListener('input', () => {
  document.getElementById('resource-result').textContent = resource.value || 'No resource entered.';
});
document.getElementById('skill-submit').addEventListener('click', () => {
  document.getElementById('skill-result').textContent = document.getElementById('skill').value.trim() || 'Enter a skill first.';
});
const age = document.getElementById('age');
function showAge() {
  const rawAge = age.value;
  // HTML input values are strings, even when the input type is number.
  document.getElementById('age-result').textContent = rawAge.trim() === '' || !age.validity.valid
    ? 'Enter a nonnegative age.'
    : 'String + 1: ' + (rawAge + 1) + '; Number(value) + 1: ' + (Number(rawAge) + 1);
}
age.addEventListener('input', showAge);
showAge();
const email = document.getElementById('practice-email');
function showEmail() {
  const value = email.value.trim();
  const valid = /^[^\s@]+@[^\s@]+\.[^\s@]+$/.test(value);
  document.getElementById('email-result').textContent = valid ? 'PASS: simple email format.' : 'FIX: use a name, @, domain, and domain suffix without spaces.';
}
email.addEventListener('input', showEmail);
showEmail();
</script>
````

## Input grammar and behavior

`ocs__input` supplies the shared appearance; `small`, `medium`, and `large` communicate size. `gradient` adds a reusable background. The name field is compact, the email field standard, and the assistance field large.

The bonus form reads strings from `.value`, uses the browser's required/email validation, and writes output with `textContent`. It does not transmit data or create a real signup. Use fictional values when trying it.

## Code runner answers

For `rawAge = "21"`, `rawAge + 1` produces `"211"`, whereas `Number(rawAge) + 1` produces `22`. Convert before arithmetic, and reject blank input first because `Number("")` is zero. A malformed number converts to `NaN`.

The email practice accepts `alex@example.com`; it rejects `alex`, `alex@`, and `alex @example.com`. Its pattern requires nonempty parts around `@`, a dot in the domain, and no whitespace. This is only a teaching check.

The browser input/output flow can also be expressed in AP CSP pseudocode:

```text
DISPLAY("Enter a volunteer skill")
skill ← INPUT()
DISPLAY(skill)
```
