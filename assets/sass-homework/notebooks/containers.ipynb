## Homework — Rotary Parade hub

The required solution uses only OCS container grammar, including the table and callout.

````python
%%html
<!-- UI_RUNNER: Homework — Rotary Parade hub -->
<div class="ocs__container">
  <article class="ocs__card">
    <h2 class="ocs__section-title">Poway Scripps Rotary Parade</h2>
    <p class="ocs__description">Plan a morning of service with this fictional parade schedule.</p>
    <div class="ocs__grid ocs__grid--standard cols-3">
      <div class="ocs__grid-cell ocs__grid-cell--header">Parade day at a glance</div>
      <div class="ocs__grid-cell">8:00 a.m. — volunteer check-in</div>
      <div class="ocs__grid-cell ocs__grid-cell--accent">9:00 a.m. — parade begins</div>
      <div class="ocs__grid-cell ocs__grid-cell--muted">11:00 a.m. — cleanup crew</div>
    </div>
    <div class="ocs__table-wrap">
      <table class="ocs__table">
        <caption>Sample volunteer assignments</caption>
        <thead><tr><th scope="col">Team</th><th scope="col">Responsibility</th><th scope="col">Time</th></tr></thead>
        <tbody>
          <tr><th scope="row">Welcome</th><td>Direct arriving volunteers</td><td>8:00 a.m.</td></tr>
          <tr><th scope="row">Cleanup</th><td>Collect signs after the parade</td><td>11:00 a.m.</td></tr>
        </tbody>
      </table>
    </div>
    <p class="ocs__callout">Check the actual organizer's information before planning a visit.</p>
  </article>
</div>
````

## Structure and responsive behavior

Each required example has exactly one container around one card. The homework adds a three-column grid with header, accent, and muted roles, a wrapped table with two body rows, and a closing callout. Below 600px the standard grid uses two columns; the table wrapper scrolls if it needs more room.
