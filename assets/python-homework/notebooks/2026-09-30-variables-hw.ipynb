## Homework

Prediction: 12 + 18 + 10 is 40. A five-star bonus leaves the stored total at 40 until recalculation makes it 45. A zero bonus keeps every total at 40.

````python
# CODE_RUNNER: 3.1 Homework - star counter and both tests
for bonus in [5, 0]:
    game_title = "Comet Garden"
    reduced_motion = True
    level_one_stars = 12
    level_two_stars = 18
    level_three_stars = 10
    total_stars = level_one_stars + level_two_stars + level_three_stars
    original_total = total_stars
    print(game_title, "| reduced motion:", reduced_motion, "| bonus:", bonus)
    print("Original total:", total_stars)
    level_two_stars = level_two_stars + bonus
    print("Stored total:", total_stars)
    total_stars = level_one_stars + level_two_stars + level_three_stars
    print("Recalculated total:", total_stars)
    assert original_total == 40 and total_stars == 40 + bonus
````
