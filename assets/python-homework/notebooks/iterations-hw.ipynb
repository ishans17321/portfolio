## Popcorn

With cutoffs 4 and 8, prediction: three safe readings, three actions, action total 19, critical found True. Reading 5 is never visited after 9 stops the loop.

````python
# CODE_RUNNER: 3.08 Popcorn - wildfire scanner
# Count safe readings before continue; count the critical action before break.
def scan(alerts, safe_cutoff, critical_cutoff):
    checked_count = 0
    safe_count = 0
    action_count = 0
    total_action_score = 0
    critical_found = False
    for alert in alerts:
        checked_count += 1
        if alert < safe_cutoff:
            safe_count += 1
            print(alert, "safe, skipping")
            continue
        action_count += 1
        total_action_score += alert
        print(alert, "action needed")
        if alert >= critical_cutoff:
            critical_found = True
            print("Critical reading: stopping scan")
            break
    report = (checked_count, safe_count, action_count, total_action_score, critical_found)
    print("Checked / skipped / actions / action total / critical:", report)
    return report

alerts = [1, 4, 2, 6, 3, 9, 5]
assert scan(alerts, 4, 8) == (6, 3, 3, 19, True)
````

## Homework

Theme: server utilization. Nine readings, warning threshold 60%, critical threshold 95%. Prediction: seven checked, three skipped, four actions totaling 313, with a critical reading found. This is an exercise using fictional measurements.

````python
# CODE_RUNNER: 3.08 Homework - server load monitor
# Checked includes all visited readings; total_score includes action readings only.
def check_server_loads(server_loads, warning_load, critical_load):
    checked_count = 0
    skipped_count = 0
    action_count = 0
    total_score = 0
    critical_found = False
    for load in server_loads:
        checked_count += 1
        if load < warning_load:
            skipped_count += 1
            print(f"Server load {load}%: safe, no action")
            continue
        action_count += 1
        total_score += load
        print(f"Server load {load}%: action needed")
        if load >= critical_load:
            critical_found = True
            print("Critical server load: stop checking and investigate.")
            break
    print("Final server report")
    print("Checked:", checked_count)
    print("Skipped:", skipped_count)
    print("Actions:", action_count)
    print("Total action load:", total_score)
    print("Critical found:", critical_found)
    return checked_count, skipped_count, action_count, total_score, critical_found
server_loads = [18, 64, 35, 82, 70, 42, 97, 50, 66]
assert check_server_loads(server_loads, 60, 95) == (7, 3, 4, 313, True)
print("Boundary test: warning included; critical included before stopping")
assert check_server_loads([59, 60, 95, 96], 60, 95) == (3, 1, 2, 155, True)
print("No critical reading:")
assert check_server_loads([20, 60], 60, 95) == (2, 1, 1, 60, False)
print("Empty input:")
assert check_server_loads([], 60, 95) == (0, 0, 0, 0, False)
````

## While-loop practice

Prediction: 70, 55, 40, 25. Decreasing the temperature eventually makes `temperature > 30` false. AP REPEAT UNTIL temperature <= 30 describes the same stopping rule.

````python
# CODE_RUNNER: 3.08 Practice - terminating while loop
# Subtracting 15 changes the condition so this while loop eventually stops.
temperature = 85
while temperature > 30:
    temperature -= 15
    print("Cooling:", temperature)
assert temperature == 25
````
