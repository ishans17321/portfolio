## Popcorn 2

Prediction: spec 1.1 is rejected as a duplicate; changing it to 1.2 passes. `and` enforces every requirement; `not` excludes existing specifications.

````python
# CODE_RUNNER: 3.05 Popcorn 2 - creation rule
# Every required field must pass, and an existing spec must block creation.
def validate_record(record, existing_spec_numbers):
    has_name = bool(record.get("product_name", "").strip())
    category_ok = record.get("category") == "Auto Racing" or record.get("category") == "Drag Racing"
    has_spec = bool(record.get("spec_number", "").strip())
    has_date = bool(record.get("effective_date", "").strip())
    duplicate = record.get("spec_number", "").strip() in existing_spec_numbers
    return has_name and category_ok and has_spec and has_date and not duplicate

part = {"product_name": "Replacement Flywheels and Clutch Assemblies", "category": "Auto Racing", "spec_number": "1.1", "effective_date": "Nov. 9, 2001"}
existing_spec_numbers = ["1.1", "2.1"]
for spec, expected in [("1.1", False), ("1.2", True)]:
    part["spec_number"] = spec
    decision = validate_record(part, existing_spec_numbers)
    print(spec, "create:", decision)
    assert decision == expected
````

## Popcorn 3

Prediction: query 1.2 matches the clutch record; flywheel matches two names. A matching name with an unaccepted category must still be excluded.

````python
# CODE_RUNNER: 3.05 Popcorn 3 - search filter
# Match a name OR an exact spec, then require an accepted category in either case.
records = [
    {"product_name": "Replacement Flywheels and Clutch Assemblies", "category": "Auto Racing", "spec_number": "1.1"},
    {"product_name": "Multiple Disc Clutch Assemblies", "category": "Drag Racing", "spec_number": "1.2"},
    {"product_name": "Racing Flywheel Record", "category": "Auto Racing", "spec_number": "2.1"}
]
def search(records, query):
    results = []
    for record in records:
        matches = query.lower() in record["product_name"].lower() or query == record["spec_number"]
        valid_category = record["category"] in ["Auto Racing", "Drag Racing"]
        if matches and valid_category:
            results.append(record)
    return results
for query in ["1.2", "flywheel", "missing"]:
    print(query, "->", search(records, query))
invalid = {"product_name": "Flywheel", "category": "Street Car", "spec_number": "9.9"}
assert search([invalid], "flywheel") == []
assert len(search(records, "1.2")) == 1
print("Invalid category excluded: verified")
````

## Homework

Predictions: accept the new clutch record, reject Street Car, reject duplicate 1.1. In the AP pseudocode, the first record also reaches ACCEPT RECORD because every required Boolean is true and duplicateSpec is false.

````python
# CODE_RUNNER: 3.05 Homework - reusable SFI validator
# Return one Boolean from named checks; blank fields and duplicate specs must fail.
def validate_record(record, existing_spec_numbers):
    has_name = bool(record.get("product_name", "").strip())
    category_ok = record.get("category") == "Auto Racing" or record.get("category") == "Drag Racing"
    has_spec = bool(record.get("spec_number", "").strip())
    has_date = bool(record.get("effective_date", "").strip())
    duplicate = record.get("spec_number", "").strip() in existing_spec_numbers
    return has_name and category_ok and has_spec and has_date and not duplicate

existing_spec_numbers = ["1.1", "2.1"]
test_records = [
    {"product_name": "Multiple Disc Clutch Assemblies", "category": "Auto Racing", "spec_number": "1.2", "effective_date": "Feb. 9, 2006"},
    {"product_name": "Replacement Flywheels", "category": "Street Car", "spec_number": "3.1", "effective_date": "Jan. 1, 2026"},
    {"product_name": "Existing Flywheel Record", "category": "Drag Racing", "spec_number": "1.1", "effective_date": "Jan. 1, 2026"}
]
for record, expected in zip(test_records, [True, False, False]):
    accepted = validate_record(record, existing_spec_numbers)
    print(record["product_name"], "->", "ACCEPT RECORD" if accepted else "REJECT RECORD")
    assert accepted == expected
for field in ["product_name", "spec_number", "effective_date"]:
    record = test_records[0].copy()
    record[field] = "   "
    assert not validate_record(record, existing_spec_numbers)
    print("Blank", field, "-> REJECT RECORD")
assert not validate_record({}, existing_spec_numbers)
print("Missing fields -> REJECT RECORD")
````
