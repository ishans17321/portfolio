## Overview

Parameters name inputs in a definition; arguments supply their values during a call. Helpers return results that an outer procedure uses.

## Popcorn Hack 2 — Decomposition and Helper Procedures

Prediction: the medical incident uses the priority branch; the power outage uses the standard branch. The helper’s returned Boolean controls the outer if, as in the lesson’s pseudocode.

````python
# CODE_RUNNER: Procedures Popcorn 2 - decomposition
# The helper returns a Boolean that selects the message in the outer procedure.
def is_priority(incident_type):
    return incident_type == "Medical Emergency"
def process_report(location, incident_type):
    priority = is_priority(incident_type)
    if priority:
        print(f"PRIORITY: {incident_type} reported in {location}")
    else:
        print(f"STANDARD REPORT: {incident_type} reported in {location}")
process_report("Poway", "Medical Emergency")
process_report("Poway", "Power Outage")
assert is_priority("Medical Emergency") is True
assert is_priority("Power Outage") is False
````

### Recorded output

```text
PRIORITY: Medical Emergency reported in Poway
STANDARD REPORT: Power Outage reported in Poway
```

## Homework Hack — Emergency-Response Program

This is a classroom simulation: its simple priority rule is not an operational dispatch protocol. The helper centralizes that rule so changing it requires one edit.

````python
# CODE_RUNNER: Procedures Homework - dispatch simulation
# Centralize the simulation priority rule in a helper instead of repeating it.
def check_priority(incident_type):
    """Return a Boolean for the lesson's fictional incident categories."""
    return incident_type == "Medical Emergency"

def dispatch_response(location, incident_type):
    priority = check_priority(incident_type)
    if priority:
        print(f"SIMULATION priority response: {incident_type} at {location}")
    else:
        print(f"SIMULATION standard response: {incident_type} at {location}")

dispatch_response("Poway", "Medical Emergency")
dispatch_response("North practice zone", "Power Outage")
dispatch_response("South practice zone", "Supply Request")
assert check_priority("Medical Emergency") is True
assert check_priority("Supply Request") is False
````

### Recorded output

```text
SIMULATION priority response: Medical Emergency at Poway
SIMULATION standard response: Power Outage at North practice zone
SIMULATION standard response: Supply Request at South practice zone
```
