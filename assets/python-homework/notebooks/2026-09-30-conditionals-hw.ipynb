## Homework

Reachability proof with the required scoring weights:

| Message | Amount | Known contact | Score | Branch |
| --- | ---: | --- | ---: | --- |
| Meeting at noon | 0 | True | 0 | else |
| Meeting at noon | 101 | True | 3 | >=3 |
| Send gift cards | 0 | False | 6 | >=6 |
| URGENT: send gift cards now to unlock your account | 500 | False | 9 | >=9 |

The separate scoring if statements need no else because a missing signal adds zero. The reporting chain does need else because every message needs a response.

````python
# CODE_RUNNER: 3.6 Homework - score and advise
def advise(message, amount_requested, known_contact):
    risk_score = 0
    # No else: when this signal is absent its contribution is zero.
    if "gift card" in message.lower():
        risk_score += 4
    # Independent signal; no else needed because zero is already accumulated.
    if amount_requested > 100:
        risk_score += 3
    # No else: a known contact contributes no unknown-sender points.
    if not known_contact:
        risk_score += 2
    print("Score:", risk_score)
    # 3 thresholds -> 4 outcomes -> 4 answers
    if risk_score >= 9:
        print("Highest risk: pause and verify through a trusted contact method.")
    elif risk_score >= 6:
        print("High risk: ask a trusted person to review the request.")
    elif risk_score >= 3:
        print("Some warning signs: verify before responding.")
    else:
        print("No scored warning signs; this does not establish authenticity.")
    return risk_score

message = "URGENT: send gift cards now to unlock your account"
amount_requested = 500
known_contact = False
assert advise(message, amount_requested, known_contact) == 9
for message, amount_requested, known_contact, expected in [
    ("Meeting at noon", 0, True, 0),
    ("Meeting at noon", 101, True, 3),
    ("Send gift cards", 0, False, 6),
    ("Meeting at noon", 100, True, 0)
]:
    assert advise(message, amount_requested, known_contact) == expected
````
