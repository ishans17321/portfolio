## Exercise 1 and mini-game

Descriptive names reveal actions and inputs. The runner uses editable variables instead of input prompts; both run and cancel paths are demonstrated. Naming a function `do_step` hides the purpose without changing its behavior.

````python
# CODE_RUNNER: Developing procedures - name and run
def choose_dessert_ingredient():
    return "oats"
def mix_ingredient_into_bowl(ingredient):
    return f"Mix {ingredient} into the bowl"
def bake_dessert():
    return "Bake the dessert"
def serve_dessert():
    return "Serve the dessert"
def prepare_oat_dessert():
    ingredient = choose_dessert_ingredient()
    return [mix_ingredient_into_bowl(ingredient), bake_dessert(), serve_dessert()]

procedure_name = "prepare_oat_dessert"
procedures = {procedure_name: prepare_oat_dessert, "do_step": prepare_oat_dessert}
for run_choice in ["run", "cancel"]:
    print("Procedure:", procedure_name, "| choice:", run_choice)
    if run_choice == "run":
        for step in procedures[procedure_name]():
            print(step)
    else:
        print("Canceled")
assert procedures["do_step"]() == procedures[procedure_name]()
print("Names change readability, not this procedure's result.")
````

## Exercise 2 — actual Java methods

The source’s first “Java” example uses JavaScript syntax. This translation uses a Java class and typed methods.

```java
public class DessertRobot {
    private static String chooseIngredient() { return "oats"; }
    private static String mixIngredient(String ingredient) {
        return "Mix " + ingredient + " into the bowl";
    }
    private static String bakeDessert() { return "Bake the dessert"; }
    private static String serveDessert() { return "Serve the dessert"; }
    public static void prepareOatDessert() {
        System.out.println(mixIngredient(chooseIngredient()));
        System.out.println(bakeDessert());
        System.out.println(serveDessert());
    }
    public static void main(String[] args) { prepareOatDessert(); }
}
```

`public` exposes the entry point to other callers; `private` keeps helpers inside the class. `static` permits calls without creating an object. Python uses naming conventions such as a leading underscore rather than Java’s enforced private access.

## Exercise 3 — decomposition and tests
