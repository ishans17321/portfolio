## Homework

Each entry is the collected stars for one level. Prediction: [12, 23, 10, 7], four levels, 52 stars; the empty project totals zero.

````python
# CODE_RUNNER: 3.2 Homework - grow and total the collection
level_stars = [12, 18, 10]
level_stars[1] += 5
level_stars.append(7)
total_stars = 0
for stars in level_stars:
    total_stars += stars
print("Level stars:", level_stars)
print("Level count:", len(level_stars))
print("Total stars:", total_stars)
assert level_stars == [12, 23, 10, 7] and total_stars == 52
````

### Recorded output

```text

Total stars: 52
```

## Tests

````python
# CODE_RUNNER: 3.2 Tests - traversal for any length
for level_stars, expected in [([10, 20, 15], 45), ([10, 20, 15, 12], 57), ([], 0)]:
    total_stars = 0
    for stars in level_stars:
        total_stars += stars
    print("Levels:", len(level_stars), "| stars:", total_stars)
    assert total_stars == expected
# No second-element access or append happens in the empty total-only test.
````
