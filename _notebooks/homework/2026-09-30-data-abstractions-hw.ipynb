## Popcorn

Warm-up prediction: 10, 20, 3; replacing the first value with 14 changes only the first result. Checkpoint prediction: [6, 12, 12, 5], length 4. Python index 1 corresponds to AP index 2.

````python
# CODE_RUNNER: 3.2 Warm-up - indexing
# Python indices start at zero; replacing one score does not change the list length.
for first in [10, 14]:
    level_stars = [first, 20, 15]
    print(level_stars[0], level_stars[1], len(level_stars))
````

### Recorded output

```text

14 20 3
```

````python
# CODE_RUNNER: 3.2 Popcorn - update and append
# Replace the second element before appending a new level at the end.
for new_level in [5, 8]:
    level_stars = [6, 9, 12]
    level_stars[1] = 12
    level_stars.append(new_level)
    print(level_stars)
    print("Levels:", len(level_stars))
    assert level_stars == [6, 12, 12, new_level]
````

### Recorded output

```text
[6, 12, 12, 5]
Levels: 4
[6, 12, 12, 8]
Levels: 4
```

Replacing an element preserves the length; appending adds one element. In Python, assigning one list name to another shares the list; use `.copy()` for an independent shallow copy. AP list assignment has different copying semantics.

## Homework

Each entry is the collected stars for one level. Prediction: [12, 23, 10, 7], four levels, 52 stars; the empty project totals zero.

````python
# CODE_RUNNER: 3.2 Homework - grow and total the collection
# Keep scores in level order and accumulate each score into a fresh total.
level_stars = [12, 18, 10]
level_stars[1] += 5
level_stars.append(7)
total_stars = 0
for stars in level_stars:
    total_stars += stars
print("Level stars:", level_stars)
print("Level count:", len(level_stars))
print("Total stars:", total_stars)
assert level_stars == [12, 23, 10, 7] and total_stars == 52
````

### Recorded output

```text
Level stars: [12, 23, 10, 7]
Level count: 4
Total stars: 52
```

## Tests

````python
# CODE_RUNNER: 3.2 Tests - traversal for any length
# The same loop handles three levels, four levels, and an empty collection.
for level_stars, expected in [([10, 20, 15], 45), ([10, 20, 15, 12], 57), ([], 0)]:
    total_stars = 0
    for stars in level_stars:
        total_stars += stars
    print("Levels:", len(level_stars), "| stars:", total_stars)
    assert total_stars == expected
# No second-element access or append happens in the empty total-only test.
````

### Recorded output

```text
Levels: 3 | stars: 45
Levels: 4 | stars: 57
Levels: 0 | stars: 0
```
