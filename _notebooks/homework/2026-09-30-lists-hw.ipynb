````python
# CODE_RUNNER: Lists Popcorn 1 - index repair
items_needed = ["blankets", "socks", "water", "meals"]
print(items_needed[0])  # blankets
print(items_needed[len(items_needed) - 1])  # meals
print(len(items_needed))  # 4
try:
    print(items_needed[4])  # predicted IndexError
except IndexError:
    print("Original fourth access: IndexError")
print("Fixed fourth access:", items_needed[3])  # meals
````

## Homework

Prediction: 12 donations, last socks, total 31, integer average 2. The exact average is 31/12; int truncates it as requested. Frequency totals are 5 blankets, 4 socks, 2 water, 1 meals.

````python
# CODE_RUNNER: Lists Homework - donation report
def donation_report(donations_by_item, costs):
    if len(donations_by_item) != len(costs):
        raise ValueError("One cost is required per donation")
    print(len(donations_by_item))
    if donations_by_item:
        print(donations_by_item[-1])
    else:
        print("No last donation")
    total = 0
    for cost in costs:
        total += cost
    print("Total:", int(total))
    if costs:
        print("Average:", int(total / len(costs)))
    else:
        print("Average: No data yet")
    frequency = {}
    for item in donations_by_item:
        if item in frequency:
            frequency[item] += 1
        else:
            frequency[item] = 1
    print(frequency)
    popular = []
    for item in frequency:
        if frequency[item] >= 3:
            popular.append(item)
    print(popular)
    return total, frequency, popular

donations_by_item = ["blankets", "socks", "blankets", "water", "socks", "blankets", "meals", "socks", "blankets", "water", "blankets", "socks"]
costs = [2, 3, 2, 4, 3, 2, 1, 3, 2, 4, 2, 3]
original = donations_by_item.copy()
result = donation_report(donations_by_item, costs)
assert result == (31, {"blankets": 5, "socks": 4, "water": 2, "meals": 1}, ["blankets", "socks"])
assert donations_by_item == original
````
