## Popcorn Hack 1 — After-School Homework Plan

The starter selects the weekday outer branch and unfinished-work inner branch, printing Finish homework. The remaining tests reach free time and the weekend path.

````python
# CODE_RUNNER: 3.7 Popcorn 1 - homework plan
# Only inspect homework status after choosing the weekday branch.
for is_weekday, homework_finished in [(True, False), (True, True), (False, False)]:
    print(is_weekday, homework_finished)
    if is_weekday:
        if homework_finished:
            print("Free time")
        else:
            print("Finish homework")
    else:
        print("Weekend plan")
````

## Popcorn Hack 2 — Repair the Student Account Check

Both original if statements were at the same indentation, so permission was checked even when logged out. Nesting permission under login prevents that path.

````python
# CODE_RUNNER: 3.7 Popcorn 2 - account guard
# Indent the permission decision inside login so logged-out users cannot reach it.
for logged_in, has_permission in [(True, False), (True, True), (False, True)]:
    print("Logged in:", logged_in, "Permission:", has_permission)
    if logged_in:
        print("Account found")
        if has_permission:
            print("Page opened")
        else:
            print("Permission needed")
    else:
        print("Log in first")
````

### Recorded output

```text

Permission needed
Logged in: True Permission: True
Account found

Logged in: False Permission: True
Log in first
```

## Popcorn Hack 3 — Lunch and Sports Practice

Lunch is checked first, then practice within that chosen branch. Two Boolean inputs give four combinations.

````python
# CODE_RUNNER: 3.7 Popcorn 3 - all lunch and practice paths
# Try all four Boolean combinations to verify every nested recommendation.
for lunch_packed in [True, False]:
    for practice_after_school in [True, False]:
        print(lunch_packed, practice_after_school)
        if lunch_packed:
            if practice_after_school:
                print("Packed lunch; bring a practice snack.")
            else:
                print("Packed lunch; regular afternoon meal plan.")
        else:
            if practice_after_school:
                print("Buy lunch and pack a practice snack.")
            else:
                print("Buy lunch; regular afternoon meal plan.")
````

## Homework Hack — High School Daily Routine Dashboard

The account guard must succeed before schedule information is shown. Homework affects free time only inside an authenticated day plan; lunch and practice choices then determine meals and sports recommendations. Every Boolean combination is tested, including logged-out cases where all schedule checks are skipped.

````python
# CODE_RUNNER: 3.7 Homework - daily routine dashboard
# The outer login guard controls access to every nested schedule decision.
student = {"logged_in": True, "homework_finished": False, "lunch_packed": True, "practice_today": True}
def daily_routine(student):
    if student["logged_in"]:
        if student["homework_finished"]:
            print("Homework complete: plan free time.")
        else:
            print("Reserve a homework block before leisure.")
        if student["lunch_packed"]:
            print("Bring your packed lunch.")
        else:
            print("Plan to buy lunch.")
        if student["practice_today"]:
            if student["homework_finished"]:
                print("Attend practice, then dinner and recovery.")
            else:
                print("Schedule homework around practice and dinner.")
            print("Pack water and a snack for practice.")
        else:
            print("No practice: plan dinner and a free evening block.")
    else:
        print("Log in first to view your day plan.")
print("Starter case:")
daily_routine(student)
for login in [False, True]:
    for homework in [False, True]:
        for lunch in [False, True]:
            for practice in [False, True]:
                case = dict(logged_in=login, homework_finished=homework, lunch_packed=lunch, practice_today=practice)
                print("Test:", case)
                daily_routine(case)
````
