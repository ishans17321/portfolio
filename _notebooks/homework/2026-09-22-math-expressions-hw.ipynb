## Popcorn Hack 1 — Basic Arithmetic Operators

**Rubric evidence: 0.2 points.** All four expressions use the supplied variables, and each computed result has a label.

| Expression | Why it fits | Expected result |
| --- | --- | --- |
| `value_a + value_b + value_c` | Total includes all three quantities. | 135 |
| `value_a - value_b` | Difference subtracts B from A in that order. | 90 |
| `value_b * value_c` | Product multiplies B and C. | 250 |
| `value_a / value_b` | Division asks how many B-sized units fit in A. | 10.0 |


```python
# CODE_RUNNER: Popcorn Hack 1 — Basic Arithmetic Operators
value_a = 100
value_b = 10
value_c = 25

# Use variables instead of hardcoded answers so changed inputs recalculate.
addition = value_a + value_b + value_c
# Subtraction is order-sensitive: A - B differs from B - A.
subtraction = value_a - value_b
multiplication = value_b * value_c
# / returns a floating-point result, even when the quotient is a whole number.
# The supplied divisor is 10, so there is no division by zero.
division = value_a / value_b
print("Total:", addition)
print("Difference:", subtraction)
print("Product:", multiplication)
print("Division:", division)
```

## Popcorn Hack 2A — Weekly and Monthly Session Costs

The prose task asks for weekly and monthly costs; the next cell also completes the booking-capacity starter shown under the same hack. Together these cover **PH2's 0.2-point criterion** without claiming extra rubric points.

**Why this is good:** the monthly expression reuses the weekly subtotal. Updating one input changes both totals automatically. Four weeks is the lesson's simplified monthly assumption.


```python
# CODE_RUNNER: Popcorn Hack 2A — Weekly and Monthly Session Costs
cost_per_session = 35
sessions_per_week = 3
weeks_per_month = 4
# Dollars/session * sessions/week gives dollars/week: 35 * 3 = 105.
weekly_cost = cost_per_session * sessions_per_week
# Reuse the subtotal rather than typing the answer 105 into the expression.
monthly_cost = weekly_cost * weeks_per_month
print(f"Weekly Cost: ${weekly_cost:.2f}")
print(f"Monthly Cost: ${monthly_cost:.2f}")
```

## Popcorn Hack 2B — Recovery Center Booking Capacity

**Why this is good:** the expressions follow the dependency chain: booked spots → remaining spots → allocation per room. The calculation is valid for the provided nonnegative booking counts, which do not exceed capacity.

`88 / 2` produces `44.0`. It represents 44 spots per room, not a fractional person. With other inputs, whole-spot allocation may require floor division and a remainder; the extension below demonstrates that distinction.


```python
# CODE_RUNNER: Popcorn Hack 2B — Recovery Center Booking Capacity
total_monthly_capacity = 100
spots_requested_per_group = 4
sessions_booked = 3
active_rooms = 2
# Repeated booking uses multiplication: 4 spots across 3 sessions = 12.
total_spots_taken = spots_requested_per_group * sessions_booked
# Subtract occupied spots from capacity, not the other way around.
remaining_capacity = total_monthly_capacity - total_spots_taken
# Divide the remaining 88 spots equally across the 2 active rooms.
users_per_room = remaining_capacity / active_rooms
print("Total Spots Taken:", total_spots_taken)
print("Remaining Capacity:", remaining_capacity)
print("Users per Room:", users_per_room)
```

## HW — Recovery Center Cost Report

**Rubric evidence: 0.4 points.** Weekly and monthly expressions satisfy the first 0.2; adding the one-time fee and dividing monthly cost by 30 satisfy the next 0.2.

**Read alongside the code:** multiplication builds recurring costs; addition includes enrollment once; division expresses an average. The extra first-month daily average makes the fee's effect visible. Formatting with `:.2f` rounds the display to cents while retaining the calculated values.

**Model limitation:** four weeks of sessions are spread across 30 calendar days. This is the assigned estimate, not an exact calendar schedule. The $14 daily average excludes enrollment; $16.50 includes it.


```python
# CODE_RUNNER: HW — Recovery Center Cost Report
# Recovery Center Cost Report
# Goal: turn a session schedule into an understandable cost estimate.
# Each named input represents one assumption that can be changed later.

cost_per_session = 35    # Dollars for ONE session, not an entire week.
sessions_per_week = 3   # Number of sessions attended each week.
weeks_per_month = 4     # Simplified billing estimate, not a calendar fact.
enrollment_fee = 75     # Charged once, when enrolling.
days_per_month = 30     # Days over which to spread the monthly estimate.

# 1. MULTIPLICATION: repeat the session charge for each weekly session.
# Units: dollars/session * sessions/week = dollars/week.
# With these inputs: 35 * 3 = 105 dollars each week.
weekly_cost = cost_per_session * sessions_per_week

# 2. REUSE A RESULT: calculate a month from the weekly subtotal.
# With these inputs: 105 * 4 = 420 dollars for 12 sessions.
# Reusing weekly_cost avoids typing 105 as a fixed answer; changing
# cost_per_session or sessions_per_week updates later calculations too.
monthly_cost = weekly_cost * weeks_per_month

# 3. ADDITION: combine recurring charges with the one-time fee.
# With these inputs: 420 + 75 = 495 dollars for the FIRST month.
# Multiplying the fee by weeks_per_month would incorrectly charge it
# four times. Later months cost monthly_cost if the schedule is unchanged.
total_cost_with_fee = monthly_cost + enrollment_fee

# 4. DIVISION: spread session costs across the assumed 30-day month.
# With these inputs: 420 / 30 = 14.0 dollars per calendar day on average.
# This is not the price of a session or an actual daily payment.
# Use / rather than // because fractional daily costs should be preserved.
average_daily_cost = monthly_cost / days_per_month

# EXTENSION: make the effect of enrollment visible in the daily estimate.
# (420 + 75) / 30 = 16.5. Using the stored total groups the addition first.
# monthly_cost + enrollment_fee / days_per_month would divide only the fee
# because division has higher precedence than addition.
first_month_daily_cost = total_cost_with_fee / days_per_month

# OUTPUT: f-strings insert calculated values into descriptive labels.
# :.2f shows exactly two decimal places, making dollars easy to read.
# It formats the display; it does not change the stored calculation.
print(f"Weekly Cost: ${weekly_cost:.2f}")
print(f"Monthly Cost (Sessions Only): ${monthly_cost:.2f}")
print(f"First Month with Enrollment Fee: ${total_cost_with_fee:.2f}")
print(f"Average Daily Cost (Sessions Only): ${average_daily_cost:.2f}")
print(f"First-Month Average Daily Cost: ${first_month_daily_cost:.2f}")
```

## Extension — Precedence, Whole Spots, and Remainders

This extra example demonstrates the lesson's `//`, `%`, and `**` operators and explains why parentheses matter. It supports understanding without replacing any required task.

The lesson calls a stack “first-in, first-out,” but a stack is **last-in, first-out**. For its RPN example `5 3 - 2 *`, pop the right operand 3 before the left operand 5; compute `(5 - 3) * 2 = 4`. Operand order matters for subtraction and division.


```python
# CODE_RUNNER: Extension — Precedence, Whole Spots, and Remainders
available_spots = 89
rooms = 2
# Floor division assigns only whole spots; modulus tracks what is left over.
spots_per_room = available_spots // rooms
extra_spots = available_spots % rooms
print("Whole Spots per Room:", spots_per_room)
print("Unallocated Extra Spots:", extra_spots)
# Conservation check: allocated spots plus remainder must recover the total.
print("Capacity Reconstructed:", spots_per_room * rooms + extra_spots)
# Parentheses change which operation happens first.
grouped_result = (5 - 3) * 2
ungrouped_result = 5 - 3 * 2
print("Grouped (5 - 3) * 2:", grouped_result)
print("Ungrouped 5 - 3 * 2:", ungrouped_result)
# ** means repeated multiplication, so 2 ** 3 = 2 * 2 * 2.
power = 2 ** 3
print("Two Cubed:", power)
```
