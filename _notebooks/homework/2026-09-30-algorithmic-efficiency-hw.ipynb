## Popcorn

Warm-up prediction: searching for 15 takes 5 linear checks versus 3 binary checks, saving 2. For target 3, linear wins 1 versus 3. Catalog growth prediction: n=8 gives 8 versus 4 worst-case checks; n=16 gives 16 versus 5.

````python
# CODE_RUNNER: 3.17 Popcorn - search growth
# Handle zero before logarithms; the formula models worst-case checks, not runtime.
import math
for target, linear_checks, binary_checks in [(15, 5, 3), (3, 1, 3)]:
    print("Target:", target, "Linear:", linear_checks, "Binary:", binary_checks, "Saved:", linear_checks - binary_checks)
for n in [8, 16, 0]:
    linear_checks = n
    binary_checks = 0 if n == 0 else math.floor(math.log2(n)) + 1
    print("Books:", n, "Linear worst case:", linear_checks, "Binary worst case:", binary_checks)
````

## Homework

Prediction: 12 books → 12 visits and 144 ordered pairs. Doubling gives 24 visits and 576 pairs because each pair dimension doubles. Empty input gives zero in every column.

````python
# CODE_RUNNER: 3.17 Homework - library recommendation
# Linear visits grow with n; ordered pairs include every combination, giving n*n.
import math
for n, expected_pairs in [(12, 144), (24, 576), (0, 0)]:
    visits = n
    pairs = n * n
    print("Books:", n, "Visits:", visits, "Ordered pairs:", pairs)
    assert pairs == expected_pairs
n = 1000
linear_checks = n
binary_checks = math.floor(math.log2(n)) + 1
print("Repeated searches of sorted catalog:", linear_checks, "linear vs", binary_checks, "binary worst-case checks")
assert binary_checks == 10
````

## Tests

Main, doubled, and empty inputs are shown above. All variables reset per iteration. For repeated searches of a sorted, directly accessible catalog, recommend binary search: 10 rather than 1,000 worst-case checks. An unsorted catalog must first be sorted; keeping an extra sorted copy also costs memory. A single early match can still favor linear search.
