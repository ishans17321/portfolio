## Popcorn Hack 1 — Indexing and slicing

```python
# CODE_RUNNER: Popcorn Hack 1 - Extract the hidden message using indexing and slicing
transmission = "xxSIGNALxxxCONFIRMEDxx77xxDELTA"

# Evidence for the 0.1 popcorn point: each result is computed from the source.
# Python includes the start index but excludes the stop: 8 - 2 = 6 letters.
word_1 = transmission[2:8]
# CONFIRMED begins at 11 and has 9 letters, so the exclusive stop is 20.
word_2 = transmission[11:20]
# Negative slicing extracts the final five characters even if the prefix grows.
word_3 = transmission[-5:]
print("Word 1:", word_1)
print("Word 2:", word_2)
print("Word 3:", word_3)
print("Total length:", len(transmission))
# Step 2 samples alternating characters; it does not preserve the full words.
print("Every other character:", transmission[::2])
# Strings are immutable: the slices create new strings, leaving transmission intact.
```

## Popcorn Hack 2 — Cleaning and searching

```python
# CODE_RUNNER: Popcorn Hack 2 - Clean and search a noisy transmission using string methods
noisy = "   ...AGENT>>>Falcon---reports: the PACKAGE has ARRIVED!!!   "

# Evidence for the 0.1 popcorn point: clean in stages so each change is visible.
# strip removes outer whitespace, preserving spaces between meaningful words.
trimmed = noisy.strip()
print("Trimmed:", repr(trimmed))
# Each replace returns a new string. Saving the result respects immutability.
# Only remove the requested punctuation; preserve the other original content.
cleaned = trimmed.replace(">", "").replace("-", "").replace("!", "")
print("Cleaned:", cleaned)
# Normalize a copy before searching so ARRIVED and arrived both match.
lowered = cleaned.lower()
print("Lowered:", lowered)
has_arrived = "arrived" in lowered
print("Contains 'arrived':", has_arrived)
# Extension: find gives the position, while count gives the number of matches.
print("Position of 'arrived':", lowered.find("arrived"))
print("Occurrences of 'arrived':", lowered.count("arrived"))
print("Original preserved:", repr(noisy))
```

## Popcorn Hack 3 — Concatenation and formatting

```python
# CODE_RUNNER: Popcorn Hack 3 - Build a decoded report using concatenation and formatting
agent_name = "Falcon"
target_word = "MIDNIGHT"
confidence = 87
fragments = ["ALPHA", "SEVEN", "DELTA"]

# Evidence for the 0.1 popcorn point: demonstrate each requested construction.
headline = "Agent " + agent_name + " intercepted: " + target_word
print(headline)
# Convert the integer explicitly because + cannot concatenate str and int.
confidence_line = "Confidence: " + str(confidence) + "%"
print(confidence_line)
# An f-string embeds the same integer without a manual conversion.
confidence_line_fstring = f"Confidence: {confidence}%"
print(confidence_line_fstring)
# join adds separators only BETWEEN fragments, avoiding an extra trailing dash.
joined_fragments = "-".join(fragments)
print("Joined fragments:", joined_fragments)
# Extension: compare both approaches to check that they communicate the same data.
print("Formatting agrees:", confidence_line == confidence_line_fstring)
```

## Homework — Crack the Final Transmission

```python
# CODE_RUNNER: Homework - Crack the Final Transmission using slicing, methods, and formatting
raw_transmission = "  2026-09-15,GHOST,THE-EAGLE-LANDS-AT-DAWN,87  "

# Splitting (0.15): strip outer padding before splitting at real CSV commas.
# This produces four fields, with no whitespace attached to the confidence.
# This simple split fits the supplied data; quoted CSV fields containing commas
# would need a CSV parser in a larger application.
segments = raw_transmission.strip().split(",")
print("Segments:", segments)

# Indexing (0.15): zero-based positions map directly to the four source fields.
# Deriving these values instead of hardcoding answers keeps the report consistent.
date = segments[0]
agent = segments[1]
message_code = segments[2]
confidence_str = segments[3]
print("Date:", date, "| Agent:", agent, "| Message code:", message_code, "| Confidence:", confidence_str)

# Slicing (0.10): [:4] selects positions 0 through 3 without a string method.
# Comparing the extracted year gives an actual Boolean confirmation.
year_check = date[:4] == "2026"
print("Year check:", year_check)

# Methods (0.15): replace decodes separators without changing message_code.
message = message_code.replace("-", " ")
# Search a lowercase copy while retaining the original case for display.
# This is a substring check, as requested; it is not a whole-word tokenizer.
found_dawn = "dawn" in message.lower()
print("Found 'dawn':", found_dawn)

# Formatting (0.15): both representations use the decoded variables.
# Pipes separate Markdown columns and the percentage sign explains the number.
markdown_row = f"| {date} | {agent} | {message} | {confidence_str}% |"
print("Markdown row:", markdown_row)
# Double braces in an f-string produce literal JSON braces.
# Confidence is unquoted (a number), and JSON requires lowercase true/false.
# Manual formatting works for these supplied fields, which contain no quotes
# or backslashes needing JSON escaping; it is not a general JSON serializer.
json_line = f'{{"date": "{date}", "agent": "{agent}", "message": "{message}", "confidence": {confidence_str}, "contains_dawn": {str(found_dawn).lower()}}}'
print("JSON line:", json_line)
```
