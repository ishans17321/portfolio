## Popcorn Hack 1 — Translate Pseudocode to Python

Prediction: total 14, average 3.5. FOR EACH becomes for; LENGTH becomes len. The total starts outside the loop.

````python
# CODE_RUNNER: 3.09 Popcorn 1 - volunteer hours
# Initialize the accumulator once, then divide the final total by the list length.
hours = [3, 5, 2, 4]
total = 0
for h in hours:
    total += h
average = total / len(hours)
print(total)
print(average)
assert (total, average) == (14, 3.5)
````

## Popcorn Hack 2 — Predict, Then Fix

Prediction before repair: 1, because > misses the exactly-full group of 12. With >= the result becomes 2.

````python
# CODE_RUNNER: 3.09 Popcorn 2 - full workshop boundary
# Exactly capacity is full too, so the repaired condition includes equality.
signed_up = [12, 7, 15, 9]
capacity = 12
before = 0
full = 0
for count in signed_up:
    if count > capacity:
        before += 1
    if count >= capacity:
        full += 1
print("Before repair:", before)
print("Full workshops:", full)
assert (before, full) == (1, 2)
````

### Recorded output

```text

Full workshops: 2
```

## Popcorn Hack 3 — Modify the Counter into a Collector

Prediction: two open workshops, Parenting Skills and Self-Sufficiency. The same index connects each name to its attendance.

````python
# CODE_RUNNER: 3.09 Popcorn 3 - collect open workshops
# Use the same index for each workshop name and its matching attendance count.
workshops = ["Self-Defense", "Parenting Skills", "Fitness", "Self-Sufficiency"]
signed_up = [12, 7, 15, 9]
capacity = 12
open_workshops = []
for i in range(len(workshops)):
    if signed_up[i] < capacity:
        open_workshops.append(workshops[i])
print("Workshops with open seats:", len(open_workshops))
print(open_workshops)
assert open_workshops == ["Parenting Skills", "Self-Sufficiency"]
````

['Parenting Skills', 'Self-Sufficiency']

## Homework Hack — Supply Closet Restock Report

Prediction: Blankets need 12, Shampoo need 10; total 22, greatest shortage Blankets. Socks at exactly its target is excluded. The fully-stocked and empty cases must avoid indexing an empty shortage list.

````python
# CODE_RUNNER: 3.09 Homework - supply restock report
# Keep item names aligned with shortages and guard the empty result before indexing.
def restock_report(items, in_stock, target):
    if not (len(items) == len(in_stock) == len(target)):
        raise ValueError("Each item needs one stock value and one target")
    # COLLECT names and positive shortages in corresponding positions.
    short_items = []
    short_amounts = []
    for i in range(len(items)):
        if in_stock[i] < target[i]:
            short_items.append(items[i])
            short_amounts.append(target[i] - in_stock[i])
    # SUM: accumulate the shortage using a loop.
    total_needed = 0
    for amount in short_amounts:
        total_needed += amount
    # MAX: retain an index so the winning name stays attached to the amount.
    largest_index = None
    for i in range(len(short_amounts)):
        if largest_index is None or short_amounts[i] > short_amounts[largest_index]:
            largest_index = i
    # COMBINE: the collected list chooses the report branch.
    if not short_items:
        print("Fully stocked!")
    else:
        print("Restock needed:")
        for i in range(len(short_items)):
            print(f"{short_items[i]}: need {short_amounts[i]}")
        print("Total items needed:", total_needed)
        print(f"Most urgent: {short_items[largest_index]} (short by {short_amounts[largest_index]})")
    return short_items, short_amounts, total_needed
items = ["Toothbrushes", "Blankets", "Socks", "Shampoo", "Notebooks"]
in_stock = [40, 8, 25, 5, 30]
target = [30, 20, 25, 15, 30]
assert restock_report(items, in_stock, target) == (["Blankets", "Shampoo"], [12, 10], 22)
print("Fully stocked test:")
assert restock_report(items, target, target) == ([], [], 0)
print("Empty closet test:")
assert restock_report([], [], []) == ([], [], 0)
````
