## Popcorn 1

Changes: jersey range 1–55; square court area 225. Prediction: a jersey in that inclusive range, side length 15, and average 24.4 rounded upward to 25. Random output is reproducible with a seed; changing it explores another draw.

````python
# CODE_RUNNER: Libraries Popcorn 1 - use documented APIs
import math
import random
print("randint API:", random.randint.__doc__)
print("ceil API:", math.ceil.__doc__)
random.seed(314)
points_per_game = [24, 18, 31, 27, 22]
selected_game = random.choice(points_per_game)
jersey = random.randint(1, 55)
side = math.sqrt(225)
average = sum(points_per_game) / len(points_per_game)
print("Game to review:", selected_game)
print("Jersey:", jersey)
print("Square court side:", side)
print("Average:", average, "Rounded upward:", math.ceil(average))
assert 1 <= jersey <= 55 and side == 15 and math.ceil(average) == 25
````

## Popcorn 2

The fictional roster is Ishan, Mira, and Dev. Added `/record`, which returns wins and losses. All routes are defined before creating the test client; each request must return HTTP 200 and the expected JSON.

````python
# CODE_RUNNER: Libraries Popcorn 2 - Flask scoreboard routes
from flask import Flask, jsonify
app = Flask("portfolio_volleyball")
@app.route("/score")
def score():
    return jsonify({"home": 23, "away": 21, "set": 2})
@app.route("/roster")
def roster():
    return jsonify({"players": ["Ishan", "Mira", "Dev"]})
@app.route("/record")
def record():
    return jsonify({"wins": 3, "losses": 1})
client = app.test_client()
for route in ["/score", "/roster", "/record"]:
    response = client.get(route)
    print(route, response.status_code, response.get_json())
    assert response.status_code == 200
assert client.get("/record").get_json() == {"wins": 3, "losses": 1}
assert client.get("/missing").status_code == 404
print("Unknown route: 404 verified")
````

## Homework

Fictional volleyball season: points are successful attacks, blocks, and serves credited to the player. Prediction: 60 total, 15 average, best match 19, with 33 against Harbor and 27 against Summit. The chosen next game is October 12, 2026; the countdown uses the actual execution date, which is printed.

````python
# CODE_RUNNER: Libraries Homework - fictional season report
import json
import statistics
from datetime import date

def summarize_season(season):
    """Summarize fictional volleyball scoring.

    Parameters:
        season: list of dictionaries with date, opponent, and numeric points.
    Returns:
        dict with total_points, average_points, best_game, points_by_opponent.
        An empty season returns zeros, an empty opponent map, and best_game None.
    """
    points = []
    points_by_opponent = {}
    for game in season:
        points.append(game["points"])
        opponent = game["opponent"]
        points_by_opponent[opponent] = points_by_opponent.get(opponent, 0) + game["points"]
    return {
        "total_points": sum(points),
        "average_points": statistics.mean(points) if points else 0,
        "best_game": max(points) if points else None,
        "points_by_opponent": points_by_opponent
    }
games = [
    {"date": "2026-09-05", "opponent": "Harbor", "points": 14},
    {"date": "2026-09-12", "opponent": "Summit", "points": 11},
    {"date": "2026-09-19", "opponent": "Harbor", "points": 19},
    {"date": "2026-09-26", "opponent": "Summit", "points": 16}
]
report = summarize_season(games)
next_game = date(2026, 10, 12)
today = date.today()
report["as_of"] = today.isoformat()
report["next_game"] = next_game.isoformat()
report["days_until_next_game"] = (next_game - today).days
print(json.dumps(report, indent=2))
assert report["total_points"] == 60 and report["average_points"] == 15
assert report["best_game"] == 19 and report["points_by_opponent"] == {"Harbor": 33, "Summit": 27}
assert summarize_season([])["best_game"] is None
# Standard-library modules are included with Python, so only external dependencies appear.
requirements_txt = "pandas\nFlask\n# json, datetime, and statistics are included with Python.\n"
print("requirements.txt:")
print(requirements_txt)
````
