## Popcorn — replace presentation-only divs

One topic heading, one supporting heading, and one lead paragraph; no custom CSS.

```python
%%html
<!-- UI_RUNNER: Popcorn — replace presentation-only divs -->
<div class="ocs_card">
  <h2 class="ocs_section-title">Meet a friendly cat</h2>
  <h3>Let the cat approach first</h3>
  <p class="ocs_lead">Move slowly and give the cat space to decide whether to interact.</p>
</div>
```

## Homework — semantic cat instructions

Includes a meaningful heading hierarchy, prose, strong emphasis, and one ordered list with exactly three items.

```python
%%html
<!-- UI_RUNNER: Homework — semantic cat instructions -->
<div class="ocs_card">
  <h2 class="ocs_section-title">How to Pat a Cat</h2>
  <h3>Have you ever wanted to pat a cat?</h3>
  <p class="ocs_lead">Follow these steps gently, and let the cat choose whether to stay.</p>
  <h4>How to pat a cat</h4>
  <p class="ocs_text"><strong>Stop if the cat moves away or seems uncomfortable.</strong></p>
  <ol>
    <li>Put out your hand slowly and let the cat approach.</li>
    <li>Gently touch the cat if it welcomes contact.</li>
    <li>Move your hand back and forth softly, checking its reaction.</li>
  </ol>
  <h3>Congrats! You pat a cat</h3>
  <p class="ocs_description">A short, calm interaction is enough.</p>
</div>
```

## Why these tags work

`h2` starts the topic; `h3` introduces subsections; `h4` names the steps inside a subsection. Paragraphs hold prose, `strong` marks an important instruction, and an ordered list preserves the sequence. OCS typography roles use a **single underscore** in this lesson (`ocs_card`, `ocs_lead`); the container lesson uses two.

The refactor removes the hand-written font rules and fake numbered divs. A screen reader can now navigate the headings and recognize a three-step list.
