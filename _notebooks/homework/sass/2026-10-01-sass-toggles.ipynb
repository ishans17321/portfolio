## Practice — connect a toggle to content

The checkbox and content have separate IDs, and a change handler connects the state to visibility.

```python
%%html
<!-- UI_RUNNER: Practice — connect a toggle to content -->
<label class="ocs__toggle ocs__toggle--green">
  <input id="notes-toggle" class="ocs__toggle-input" type="checkbox">
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Show flight notes</span>
</label>
<p id="flight-notes" hidden>Check the control surfaces before the sample flight.</p>
<script>
const control = document.getElementById('notes-toggle');
const notes = document.getElementById('flight-notes');
control.addEventListener('change', () => { notes.hidden = !control.checked; });
</script>
```

## Homework — interactive flight project settings

Three labeled toggles have different working purposes. The count initializes to 1 and updates through every combination.

```python
%%html
<!-- UI_RUNNER: Homework — interactive flight project settings -->
<section aria-label="Project settings">
<label class="ocs__toggle ocs__toggle--green">
  <input id="details-toggle" class="ocs__toggle-input" type="checkbox" checked>
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Show project notes</span>
</label>
<label class="ocs__toggle ocs__toggle--green">
  <input id="theme-toggle" class="ocs__toggle-input" type="checkbox">
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Light theme</span>
</label>
<label class="ocs__toggle ocs__toggle--green">
  <input id="compact-toggle" class="ocs__toggle-input" type="checkbox">
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Compact cards</span>
</label>
<p id="enabled-count" role="status" aria-live="polite"></p>
</section>
<section class="ocs__card">
  <h2 class="ocs__section-title">RC flight project</h2>
  <p class="ocs__description">An interface demo inspired by my RC planes and drones.</p>
  <div id="project-notes">
    <h3>Project notes</h3>
    <ul><li>Inspect the airframe.</li><li>Check battery level.</li><li>Review the practice route.</li></ul>
  </div>
</section>
<script>
const details = document.getElementById('details-toggle');
const theme = document.getElementById('theme-toggle');
const compact = document.getElementById('compact-toggle');
const settings = [details, theme, compact];
function updateSettings() {
  // .checked is a boolean representing the live state, unlike the initial HTML attribute.
  document.getElementById('project-notes').hidden = !details.checked;
  document.body.classList.toggle('light-mode', theme.checked);
  document.body.classList.toggle('compact-mode', compact.checked);
  // Recount the state on every change instead of incrementing a potentially stale counter.
  const enabled = settings.filter(input => input.checked).length;
  document.getElementById('enabled-count').textContent = `${enabled} of ${settings.length} toggles on`;
}
settings.forEach(input => input.addEventListener('change', updateSettings));
// Initialize both the visible features and the count before any user interaction.
updateSettings();
</script>
```

## Popcorn — fill in the blanks

1. **Checkbox:** stores whether the setting is enabled.
2. **Slider:** the visible track reflects the state.
3. **Alignment:** the switch wrapper aligns the control and label.
4. **border-radius:** rounds the track and thumb.
5. **White circle:** the `::before` pseudo-element draws the thumb. It is a pseudo-element, not a class.

## How the homework works

JavaScript reads each checkbox's boolean `.checked` property. The detail setting shows or hides the project notes; light mode changes the preview's palette; compact mode changes card spacing. Every change also recomputes the enabled count, so the count cannot drift out of sync. Labels stay clickable, native checkboxes support keyboard input, and the decorative track is hidden from assistive technology.
