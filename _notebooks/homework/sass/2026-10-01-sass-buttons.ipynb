## Popcorn 1 — base class

The base button class replaces the missing styling hook.

```python
%%html
<!-- UI_RUNNER: Popcorn 1 — base class -->
<button type="button" class="ocs__btn" data-message="Daily check-in demo opened.">Start Daily Check-In</button>
<p id="button-result" role="status">No demo action selected.</p>
<script>
// This classroom demo provides feedback without storing or sending personal data.
document.querySelectorAll('button[data-message]').forEach(button => {
  button.addEventListener('click', () => {
    document.getElementById('button-result').textContent = button.dataset.message;
  });
});
</script>
```

## Popcorn 2 — modifiers and group

The primary action has both pill and fill; both actions share one links container.

```python
%%html
<!-- UI_RUNNER: Popcorn 2 — modifiers and group -->
<div class="ocs__links">
  <button type="button" class="ocs__btn pill fill" data-message="Sample check-in completed.">Complete Daily Check-In</button>
  <button type="button" class="ocs__btn outline" data-message="Sample draft saved for this preview only.">Save Draft Journal</button>
</div>
<p id="button-result" role="status">No demo action selected.</p>
<script>
// This classroom demo provides feedback without storing or sending personal data.
document.querySelectorAll('button[data-message]').forEach(button => {
  button.addEventListener('click', () => {
    document.getElementById('button-result').textContent = button.dataset.message;
  });
});
</script>
```

## Homework 2 — primary and secondary actions

The filled primary action stands out from the outlined secondary and alert actions. The links preserve the lesson’s resource destination.

```python
%%html
<!-- UI_RUNNER: Homework 2 — primary and secondary actions -->
<div class="ocs__links">
  <button type="button" class="ocs__btn accent fill" data-message="Sample mood entry logged in this preview.">Log Mood Entry</button>
  <a class="ocs__btn outline" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener">View Weekly Trends</a>
  <a class="ocs__btn alert-red outline" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener">Crisis Hotline</a>
</div>
<p id="button-result" role="status">No demo action selected.</p>
<script>
// This classroom demo provides feedback without storing or sending personal data.
document.querySelectorAll('button[data-message]').forEach(button => {
  button.addEventListener('click', () => {
    document.getElementById('button-result').textContent = button.dataset.message;
  });
});
</script>
```

## Design explanation

The base class gives every action consistent spacing. `pill` changes its shape, `fill` gives an action visual weight, and `outline` supports a secondary action. Semantic tone classes express urgency without hardcoding colors into markup. Text labels communicate intent even when colors are difficult to distinguish.

The second homework follows the written requirement **Crisis Hotline** where the starter inconsistently says Mental Health Blog. Resource links use a normal URL rather than Markdown embedded inside an HTML href.
