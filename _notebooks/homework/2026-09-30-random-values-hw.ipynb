## Popcorn 1

Prediction: each number lies in 1–4 inclusive; repetitions are valid because selection is with replacement. The changed record count 6 extends the range to 1–6. AP RANDOM(1, 4) corresponds to randint(1, 4), not an index starting at zero.

````python
# CODE_RUNNER: Random Popcorn 1 - record IDs
# Choose inclusive record IDs; seeding repeats the demonstration, not a unique draw.
import random
for record_count in [4, 6]:
    random.seed(315)
    numbers = []
    for _ in range(12):
        record_number = random.randint(1, record_count)
        numbers.append(record_number)
        assert 1 <= record_number <= record_count
    print("Record count:", record_count, "Selections:", numbers)
````

## Popcorn 3

Possible actions are GET/search, POST/create, and PUT/update. The selected action chooses a distinct message. The second phase explicitly exercises each action, in addition to random draws.

````python
# CODE_RUNNER: Random Popcorn 3 - simulated API actions
# Vary records and actions randomly, then test each action directly for coverage.
import random
parts = [
    {"product_name": "Replacement Flywheels", "spec_number": "1.1"},
    {"product_name": "Multiple Disc Clutch Assemblies", "spec_number": "1.2"},
    {"product_name": "Racing Flywheel Record", "spec_number": "2.1"}
]
actions = ["GET/search", "POST/create", "PUT/update"]
def describe_action(part, action):
    if action == "GET/search":
        result = "Search for matching spec"
    elif action == "POST/create":
        result = "Validate candidate before creation"
    elif action == "PUT/update":
        result = "Look up existing spec before updating"
    else:
        raise ValueError("Unknown action")
    print(action, part["product_name"], part["spec_number"], "->", result)
random.seed(315)
for _ in range(5):
    describe_action(random.choice(parts), random.choice(actions))
print("Explicit path checks:")
for action in actions:
    describe_action(parts[0], action)
````

## Homework

This simulator changes only a local set of spec numbers. Search/update can find or miss a record; create adds a new spec or rejects a duplicate; delete removes an existing spec or reports it missing. Independent random selections cover three records and four actions; explicit tests verify every result branch.

````python
# CODE_RUNNER: Random Homework - reusable SFI QA simulator
# Copy the initial database into a local set; creation must reject duplicate specs.
import random
parts = [
    {"product_name": "Replacement Flywheels and Clutch Assemblies", "category": "Auto Racing", "spec_number": "1.1"},
    {"product_name": "Multiple Disc Clutch Assemblies", "category": "Drag Racing", "spec_number": "1.2"},
    {"product_name": "Racing Flywheel Record", "category": "Auto Racing", "spec_number": "2.1"}
]
actions = ["GET/search", "POST/create", "PUT/update", "DELETE/remove"]

def simulate_action(part, action, existing):
    spec = part["spec_number"]
    if action == "GET/search":
        return "FOUND" if spec in existing else "NOT FOUND"
    elif action == "POST/create":
        if spec in existing:
            return "REJECT DUPLICATE"
        existing.add(spec)
        return "CREATED"
    elif action == "PUT/update":
        return "UPDATED" if spec in existing else "NOT FOUND"
    elif action == "DELETE/remove":
        if spec not in existing:
            return "NOT FOUND"
        existing.remove(spec)
        return "REMOVED"
    raise ValueError("Unsupported action")

def run_qa(parts, actions, existing_spec_numbers, test_count, seed):
    rng = random.Random(seed)
    existing = set(existing_spec_numbers)
    for run in range(1, test_count + 1):
        part = rng.choice(parts)
        action = rng.choice(actions)
        result = simulate_action(part, action, existing)
        print(run, part["product_name"], part["category"], part["spec_number"], action, "->", result)
    print("Final simulated specs:", sorted(existing))
    return existing

existing_spec_numbers = ["1.1", "2.1"]
test_count = 5
for seed in [315, 42]:
    print("Seed:", seed)
    run_qa(parts, actions, existing_spec_numbers, test_count, seed)
# Each directed test starts from a known local database.
existing = set(existing_spec_numbers)
for index, action, expected in [
    (0, "GET/search", "FOUND"), (1, "GET/search", "NOT FOUND"),
    (0, "POST/create", "REJECT DUPLICATE"), (1, "POST/create", "CREATED"),
    (1, "POST/create", "REJECT DUPLICATE"), (1, "PUT/update", "UPDATED"),
    (1, "DELETE/remove", "REMOVED"), (1, "DELETE/remove", "NOT FOUND"),
    (1, "PUT/update", "NOT FOUND")
]:
    actual = simulate_action(parts[index], action, existing)
    print("Directed test:", action, parts[index]["spec_number"], actual)
    assert actual == expected
assert existing_spec_numbers == ["1.1", "2.1"]
print("All result branches verified; original input unchanged.")
````
